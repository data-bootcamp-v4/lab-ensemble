```python
# Libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    BaggingClassifier,
    RandomForestClassifier,
    GradientBoostingClassifier,
    AdaBoostClassifier
)
```

```python
spaceship = pd.read_csv(
    "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/spaceship_titanic.csv"
)

display(spaceship.head())
print("Shape:", spaceship.shape)
print("\nMissing values:")
display(spaceship.isna().sum())
```

## Feature engineering, feature selection and scaling

We will prepare the data in the same way for all models so that the comparison is fair.

### Decisions

- `PassengerId` and `Name` are identifiers, so we remove them.
- From `Cabin`, we extract the deck and the side because these parts may contain useful information.
- Missing numerical values are filled with the median.
- Missing categorical values are filled with the mode.
- Categorical variables are converted into numerical dummy variables.
- Numerical variables are scaled with `StandardScaler`.

Tree-based ensemble models do not strictly require scaling, but we keep it because the exercise asks us to apply the same preprocessing used in the previous lab.


```python
# Create a copy so that the original DataFrame remains unchanged
df = spaceship.copy()

# -----------------------------
# 1. Feature engineering
# -----------------------------

# Cabin has the format: Deck/Number/Side, for example B/0/P
df["CabinDeck"] = df["Cabin"].str.split("/").str[0]
df["CabinSide"] = df["Cabin"].str.split("/").str[2]

# Total amount spent by each passenger
spending_columns = [
    "RoomService",
    "FoodCourt",
    "ShoppingMall",
    "Spa",
    "VRDeck"
]

df["TotalSpending"] = df[spending_columns].sum(axis=1)

# -----------------------------
# 2. Feature selection
# -----------------------------

# Drop identifiers and the original Cabin column
df = df.drop(columns=["PassengerId", "Name", "Cabin"])

# -----------------------------
# 3. Handle missing values
# -----------------------------

numerical_columns = df.select_dtypes(include="number").columns
categorical_columns = df.select_dtypes(include=["object", "bool"]).columns

# Do not include the target in the list of categorical predictors
categorical_features = [
    column for column in categorical_columns
    if column != "Transported"
]

for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())

for column in categorical_features:
    df[column] = df[column].fillna(df[column].mode()[0])

# -----------------------------
# 4. Separate X and y
# -----------------------------

X = df.drop(columns="Transported")
y = df["Transported"].astype(int)

# Convert categorical variables into dummy variables
X = pd.get_dummies(X, drop_first=True, dtype=int)

print("Shape after preprocessing:", X.shape)
display(X.head())
```

## Train-test split

We divide the dataset into:

- **80% training data:** used by the models to learn.
- **20% test data:** used to evaluate the models on unseen observations.

`stratify=y` keeps approximately the same proportion of transported and non-transported passengers in both datasets.


```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Fit the scaler only with the training data to avoid data leakage
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Training shape:", X_train_scaled.shape)
print("Test shape:", X_test_scaled.shape)

print("\nTarget distribution in training data:")
display(y_train.value_counts(normalize=True))
```

## Model selection

We will train and compare four ensemble techniques:

1. **Bagging and Pasting**
2. **Random Forest**
3. **Gradient Boosting**
4. **AdaBoost**

We will evaluate them using:

- Accuracy
- Precision
- Recall
- F1-score
- ROC-AUC

Because the target is quite balanced, accuracy is useful. However, F1-score and ROC-AUC give us a more complete comparison.


### Bagging and Pasting

Both methods train several decision trees.

- **Bagging:** each tree receives a random sample **with replacement**.
- **Pasting:** each tree receives a random sample **without replacement**.

The final classification is obtained through majority voting.


```python
# Dictionary where we will store every model's metrics
results = []

def evaluate_model(model_name, model, X_train_data, X_test_data):
    """Train a model, calculate predictions and store its metrics."""

    model.fit(X_train_data, y_train)

    y_pred = model.predict(X_test_data)
    y_probability = model.predict_proba(X_test_data)[:, 1]

    metrics = {
        "Model": model_name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1-score": f1_score(y_test, y_pred),
        "ROC-AUC": roc_auc_score(y_test, y_probability)
    }

    results.append(metrics)

    print(f"{model_name}")
    print("-" * len(model_name))
    print(classification_report(y_test, y_pred))
    print("Confusion matrix:")
    display(pd.DataFrame(
        confusion_matrix(y_test, y_pred),
        index=["Actual 0", "Actual 1"],
        columns=["Predicted 0", "Predicted 1"]
    ))

    return model


# BAGGING: sampling with replacement
bagging_model = BaggingClassifier(
    estimator=DecisionTreeClassifier(
        max_depth=8,
        random_state=42
    ),
    n_estimators=100,
    max_samples=0.80,
    bootstrap=True,
    random_state=42,
    n_jobs=-1
)

bagging_model = evaluate_model(
    "Bagging",
    bagging_model,
    X_train_scaled,
    X_test_scaled
)


# PASTING: sampling without replacement
pasting_model = BaggingClassifier(
    estimator=DecisionTreeClassifier(
        max_depth=8,
        random_state=42
    ),
    n_estimators=100,
    max_samples=0.80,
    bootstrap=False,
    random_state=42,
    n_jobs=-1
)

pasting_model = evaluate_model(
    "Pasting",
    pasting_model,
    X_train_scaled,
    X_test_scaled
)
```

### Random Forest

Random Forest is a bagging method made specifically from decision trees.

It introduces randomness in two places:

- Each tree trains with a different sample of rows.
- Each split considers only a random subset of features.

This makes the trees more independent and usually reduces overfitting compared with one individual decision tree.


```python
random_forest_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=None,
    min_samples_leaf=2,
    max_features="sqrt",
    random_state=42,
    n_jobs=-1
)

random_forest_model = evaluate_model(
    "Random Forest",
    random_forest_model,
    X_train_scaled,
    X_test_scaled
)
```

### Gradient Boosting

Gradient Boosting trains models sequentially.

Each new tree tries to correct the errors made by the previous ensemble. The final prediction is the sum of many small corrections.

Important hyperparameters:

- `n_estimators`: number of trees.
- `learning_rate`: how much each new tree contributes.
- `max_depth`: complexity of each individual tree.

A smaller learning rate normally requires more estimators.


```python
gradient_boosting_model = GradientBoostingClassifier(
    n_estimators=150,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

gradient_boosting_model = evaluate_model(
    "Gradient Boosting",
    gradient_boosting_model,
    X_train_scaled,
    X_test_scaled
)
```

### Adaptive Boosting

AdaBoost also trains weak learners sequentially.

After each step, it gives more importance to observations that were incorrectly classified, so that the next estimator focuses more strongly on difficult cases.


```python
ada_boost_model = AdaBoostClassifier(
    n_estimators=150,
    learning_rate=0.50,
    random_state=42
)

ada_boost_model = evaluate_model(
    "AdaBoost",
    ada_boost_model,
    X_train_scaled,
    X_test_scaled
)
```

## Model comparison

The best model should not be selected only because it has the highest accuracy.

We should compare several metrics:

- **Accuracy:** overall proportion of correct predictions.
- **Precision:** among predicted transported passengers, how many were actually transported.
- **Recall:** among passengers who were transported, how many the model identified.
- **F1-score:** balance between precision and recall.
- **ROC-AUC:** ability to separate both classes across different probability thresholds.

For this dataset, the target is reasonably balanced, so we can mainly compare **F1-score**, **ROC-AUC**, and accuracy together.


```python
results_df = (
    pd.DataFrame(results)
    .sort_values(by="F1-score", ascending=False)
    .reset_index(drop=True)
)

display(results_df.style.format({
    "Accuracy": "{:.3f}",
    "Precision": "{:.3f}",
    "Recall": "{:.3f}",
    "F1-score": "{:.3f}",
    "ROC-AUC": "{:.3f}"
}))

best_model = results_df.iloc[0]

print(
    f"The best model according to F1-score is {best_model['Model']} "
    f"with an F1-score of {best_model['F1-score']:.3f}, "
    f"an accuracy of {best_model['Accuracy']:.3f}, "
    f"and a ROC-AUC of {best_model['ROC-AUC']:.3f}."
)

print(
    "\nWhy? Because it provides the best balance between precision and recall "
    "among the tested models. We should also verify that its ROC-AUC and "
    "accuracy are competitive before making the final decision."
)
```

### Final interpretation

The exact winner can vary slightly depending on the train-test split and hyperparameters. In many runs of this dataset, Gradient Boosting or Random Forest performs particularly well. The correct conclusion is therefore based on the metrics produced by your own execution, not on assuming beforehand that one algorithm must always win.
